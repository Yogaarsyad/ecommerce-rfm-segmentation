```python
import pandas as pd
import datetime as dt

# 1. Membaca 3 file dataset utama (karena filenya ada di luar, kita langsung panggil namanya)
df_orders = pd.read_csv('olist_orders_dataset.csv')
df_payments = pd.read_csv('olist_order_payments_dataset.csv')
df_customers = pd.read_csv('olist_customers_dataset.csv')

# 2. Filter: Hanya mengambil kolom yang kita butuhkan agar prosesnya cepat
df_orders = df_orders[['order_id', 'customer_id', 'order_purchase_timestamp']]
df_payments = df_payments[['order_id', 'payment_value']]
df_customers = df_customers[['customer_id', 'customer_unique_id']]

# 3. Menggabungkan (Merge) ketiga tabel menggunakan fungsi join/merge
# Gabung orders dan customers
df_merge_1 = pd.merge(df_orders, df_customers, on='customer_id')
# Gabung hasilnya dengan payments
df_final = pd.merge(df_merge_1, df_payments, on='order_id')

# 4. Mengubah tipe data tanggal (dari teks menjadi format Datetime yang bisa dihitung)
df_final['order_purchase_timestamp'] = pd.to_datetime(df_final['order_purchase_timestamp'])

# Tampilkan hasil gabungan
print(f"Total baris data gabungan: {df_final.shape[0]}")
display(df_final.head())
```
